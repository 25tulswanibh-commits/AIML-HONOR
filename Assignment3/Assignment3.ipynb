```python
import sqlite3

class Employee:

    def __init__(self):
        self.db = sqlite3.connect("employee.db")
        self.c = self.db.cursor()

        self.c.execute(
            "CREATE TABLE IF NOT EXISTS employee(id, name, department, salary)"
        )
        self.db.commit()

    def add(self):
        a = input("Enter ID: ")
        b = input("Enter Name: ")
        c = input("Enter Department: ")
        d = input("Enter Salary: ")

        self.c.execute(
            "INSERT INTO employee VALUES(?,?,?,?)",
            (a, b, c, d)
        )
        self.db.commit()
        print("Record added")

    def show(self):
        self.c.execute("SELECT * FROM employee")

        for x in self.c.fetchall():
            print(x)

    def change(self):
        a = input("Enter ID: ")
        b = input("Enter new salary: ")

        self.c.execute(
            "UPDATE employee SET salary=? WHERE id=?",
            (b, a)
        )
        self.db.commit()
        print("Record updated")

    def remove(self):
        a = input("Enter ID: ")

        self.c.execute(
            "DELETE FROM employee WHERE id=?",
            (a,)
        )
        self.db.commit()
        print("Record deleted")


obj = Employee()

while True:
    print("\n1. Add")
    print("2. Show")
    print("3. Update")
    print("4. Delete")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        obj.add()

    elif choice == "2":
        obj.show()

    elif choice == "3":
        obj.change()

    elif choice == "4":
        obj.remove()

    elif choice == "5":
        print("Thank you")
        break

    else:
        print("Wrong choice")
```
